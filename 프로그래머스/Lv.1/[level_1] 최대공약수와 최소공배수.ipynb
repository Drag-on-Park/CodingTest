### 문제 설명
두 수를 입력받아 두 수의 최대공약수와 최소공배수를 반환하는 함수, solution을 완성해 보세요. 배열의 맨 앞에 최대공약수, 그다음 최소공배수를 넣어 반환하면 됩니다. 예를 들어 두 수 3, 12의 최대공약수는 3, 최소공배수는 12이므로 solution(3, 12)는 [3, 12]를 반환해야 합니다.

### 제한 사항
두 수는 1이상 1000000이하의 자연수입니다.


```python
def solution(n, m):
    answer = []
    # 최대공약수 = n의 약수중 12로 나눌 수 있는 가장 큰수
    # 최소공배수 공통된 지수 최대치끼리 곱함 없으면 그냥 곱
    mod_l = []
    for i in range(1,n+1):
        if n % i == 0:
            mod_l.append(i)

    max_j = []
    for j in mod_l:
        if m % j == 0:
            max_j.append(j)
    print(max_j)
    if not max_j:
        answer = [1]
        answer.append(n*m)
    else:
        answer = [max(max_j)]
        n_mod = n // answer[0]
        m_mod = m // answer[0] 
        lcm = answer[0]*n_mod*m_mod
        answer.append(lcm)
    
    # # 12 / 18
    # 6
    # 2
    # 3
    # # 16 / 24
    # 8
    # 2
    # 3

            
        
        

    return answer

n = 3
m = 12
# return	[3, 12]
n = 2
m = 5
# return	[1, 10]
solution(n, m)
```
